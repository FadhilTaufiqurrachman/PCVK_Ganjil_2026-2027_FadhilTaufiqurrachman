Manakah dari ketiga kanal warna dasar (Biru, Hijau, atau Merah) yang memiliki nilai rata-rata intensitas piksel tertinggi pada keseluruhan foto KTM?

Rancangan Program

Program ini menggunakan pemotongan matriks (array slicing) untuk memisahkan setiap kanal warna BGR OpenCV, kemudian menghitung nilai rata-rata (mean) intensitas dari masing-masing kanal menggunakan fungsi statistik NumPy.

```python
# 1. Membaca Citra KTM
img_ktp = cv.imread('/content/drive/MyDrive/Education/Current Semester College File /PCVK Image/ktm4crop.jpg')

# 2. Mengambil Tiap Kanal Warna (Indeks 0=Biru, 1=Hijau, 2=Merah)
kanal_biru = img_ktp[:, :, 0]
kanal_hijau = img_ktp[:, :, 1]
kanal_merah = img_ktp[:, :, 2]

# 3. Menghitung Rata-Rata Intensitas Pada Masing-Masing Kanal
rata_biru = np.mean(kanal_biru)
rata_hijau = np.mean(kanal_hijau)
rata_merah = np.mean(kanal_merah)

# 4. Mencetak Hasil Perhitungan Nilai Rata-Rata Kanal Warna
print(f"Rata-Rata Intensitas Kanal Biru  : {rata_biru:.3f}")
print(f"Rata-Rata Intensitas Kanal Hijau : {rata_hijau:.3f}")
print(f"Rata-Rata Intensitas Kanal Merah : {rata_merah:.3f}")
```

Kesimpulan

Berdasarkan pengolahan data numerik matriks piksel, kanal warna Biru memiliki nilai rata-rata tertinggi dibandingkan kanal Hijau dan Merah. Angka ini membuktikan bahwa foto KTM didominasi oleh unsur warna biru dan latar belakang terang (mendekati nilai maksimal 255), yang konsisten dengan identitas visual kartu mahasiswa yang didominasi corak warna biru.

**Tugas Praktikum - Aplikasi**

Studi Kasus 1

```python
def Beri_Label_Identitas(img, teks):
    # Ambil Dimensi Tinggi Dan Lebar Secara Aman
    Tinggi, Lebar = img.shape[:2]
    # Buat Salinan Array Citra Agar Tampilan Asli Terlindungi
    Img_Label = img.copy()
    # Hitung Rasio Tinggi Bar Label Yaitu 1/12 Dari Total Tinggi Gambar
    Tinggi_Bar = Tinggi // 12
    # Blok Semua Baris Dari Titik Awal Bar Hingga Akhir, Ubah Menjadi Hitam
    Img_Label[Tinggi - Tinggi_Bar : Tinggi, :] = (0, 0, 0)
    # Hitung Skala Font Awal Berdasarkan Tinggi Bar
    Skala_Font = Tinggi_Bar / 40
    Tebal_Teks = max(1, int(Skala_Font * 2))
    # Ambil Ukuran Teks Untuk Mengecek Lebar Aktualnya
    (Lebar_Teks, Tinggi_Teks), _ = cv.getTextSize(teks, cv.FONT_HERSHEY_SIMPLEX, Skala_Font, Tebal_Teks)
    # Hitung Margin Kiri Agar Terlihat Rapi Dan Proporsional
    Margin_Kiri = Lebar // 40
    # Cek Jika Total Panjang Teks Ditambah Margin Kiri Dan Kanan Melebihi Batas Lebar Citra
    if (Lebar_Teks + Margin_Kiri * 2) > Lebar:
        # Hitung Ulang Skala Berdasarkan Batas Maksimal Lebar Yang Tersedia Di Kanvas
        Faktor_Koreksi = (Lebar - Margin_Kiri * 2) / Lebar_Teks
        Skala_Font = Skala_Font * Faktor_Koreksi
        # Perbarui Ketebalan Serta Tinggi Teks Sesuai Skala Font Yang Baru
        Tebal_Teks = max(1, int(Skala_Font * 2))
        (Lebar_Teks, Tinggi_Teks), _ = cv.getTextSize(teks, cv.FONT_HERSHEY_SIMPLEX, Skala_Font, Tebal_Teks)
    # Tetapkan Koordinat X Sesuai Margin Dan Posisi Y Tepat Di Sumbu Tengah Pita Hitam
    Posisi_X = Margin_Kiri
    Posisi_Y = Tinggi - (Tinggi_Bar // 2) + (Tinggi_Teks // 2)
    # Rendering Teks Menggunakan Konfigurasi Yang Telah Dikalkulasi Dan Dikalibrasi
    cv.putText(Img_Label, teks, (Posisi_X, Posisi_Y), cv.FONT_HERSHEY_SIMPLEX,
               Skala_Font, (255, 255, 255), Tebal_Teks, cv.LINE_AA)
    return Img_Label

# Membaca Citra
Img_1 = cv.imread('/content/drive/MyDrive/Education/Current Semester College File /PCVK Image/landscape.jpg')
Img_2 = cv.imread('/content/drive/MyDrive/Education/Current Semester College File /PCVK Image/potrait.jpg')
# Terapkan Fungsi Pelabelan Pada Kedua Gambar
Hasil_1 = Beri_Label_Identitas(Img_1, 'Dokumentasi PCVK - 2026')
Hasil_2 = Beri_Label_Identitas(Img_2, 'Dokumentasi PCVK - 2026')
# Siapkan Kanvas Visualisasi Matplotlib
plt.figure(figsize=(10, 6))
# Konversi Warna BGR Ke RGB Dan Tampilkan Hasil Lanskap Di Kiri
plt.subplot(1, 2, 1)
plt.imshow(cv.cvtColor(Hasil_1, cv.COLOR_BGR2RGB))
plt.title(f'Masukan {Img_1.shape[1]} x {Img_1.shape[0]} px')
# Konversi Warna BGR Ke RGB Dan Tampilkan Hasil Potret Di Kanan
plt.subplot(1, 2, 2)
plt.imshow(cv.cvtColor(Hasil_2, cv.COLOR_BGR2RGB))
plt.title(f'Masukan {Img_2.shape[1]} x {Img_2.shape[0]} px')

plt.show()
```

```python
def buat_thumbnail(img):
    # Ambil Dimensi Asli Citra
    h, w = img.shape[:2]
    # Tentukan Ukuran Kanvas Target Dan Skala Pengecilan
    sisi = 800
    skala = sisi / max(h, w)
    # Hitung Ukuran Baru Kemudian Ubah Ukuran Citra
    h_baru, w_baru = int(h * skala), int(w * skala)
    img_kecil = cv.resize(img, (w_baru, h_baru), interpolation=cv.INTER_AREA)
    # Buat Kanvas Putih Kosong Tiga Kanal
    kanvas = np.full((sisi, sisi, 3), 255, dtype=np.uint8)
    # Hitung Posisi Tengah Kanvas
    y, x = (sisi - h_baru) // 2, (sisi - w_baru) // 2
    # Tempel Citra Mengecil Tepat Di Tengah Kanvas
    kanvas[y : y + h_baru, x : x + w_baru] = img_kecil
    # Hitung Tinggi Bar Bawah Sesuai Ketentuan Soal
    bar = sisi // 12
    # Warnai Bagian Bawah Kanvas Menjadi Hitam Dengan Slicing
    kanvas[sisi - bar : sisi, :] = 0
    # Tulis Teks Rata Kiri Dengan Koordinat Dan Ukuran Adaptif
    cv.putText(kanvas, 'Toko Saya', (sisi // 30, sisi - (bar // 3)),
               cv.FONT_HERSHEY_SIMPLEX, bar / 40, (255, 255, 255),
               max(1, int(bar / 20)), cv.LINE_AA)
    return kanvas

# Baca Citra Dari Drive
img_lanskap = cv.imread('/content/drive/MyDrive/Education/Current Semester College File /PCVK Image/landscape.jpg')
img_potret = cv.imread('/content/drive/MyDrive/Education/Current Semester College File /PCVK Image/potrait.jpg')
# Buat Kanvas Figure Matplotlib
plt.figure(figsize=(10, 5))
# Konversi Warna Dan Tampilkan Hasil Lanskap Di Subplot Kiri
plt.subplot(1, 2, 1)
plt.imshow(cv.cvtColor(buat_thumbnail(img_lanskap), cv.COLOR_BGR2RGB))
# Konversi Warna Dan Tampilkan Hasil Potret Di Subplot Kanan
plt.subplot(1, 2, 2)
plt.imshow(cv.cvtColor(buat_thumbnail(img_potret), cv.COLOR_BGR2RGB))

plt.show()
```
