1. Import Library

```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow # For Image Display
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt
```

2. Membaca dan menampilkan image

```python
# Membuat List Untuk Menyimpan Url Dari Beberapa Image
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
        "https://placekittens.com/800/571"]

# Baca Dan Tampilkan Image
# Loop Pada Tiap Url Image, Beberapa Image Dapat Disimpan Pada List
for url in urls:
    image = io.imread(url)                                # Read Image
    image = cv.resize(image, (0, 0), fx=0.5, fy=0.5)      # Resize Image To Half Size
    image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)        # Convert Color To RGB
    final_frame = cv.hconcat((image, image_2))            # Concatenate Image
    cv2_imshow(final_frame)                               # Show Image
    print('\n')
```

3. Melihat ukuran file image

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("Resolusi Image: Tinggi X Lebar = ", tinggi, " X ", lebar)
cv2_imshow(image_2)
```

4. Mengakses pixel dengan memberikan
garis horizontal berwarna putih di tengah image.

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# Membuat Garis Horizontal Di Tengah Image
for y in range(lebar):
    image_3[int((tinggi) / 2), y] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

2. Membuat garis vertikal dan garis menyilang diagonal pada image keluaran.

Garis Vertikal :

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# Membuat Garis Vertikal Di Tengah Image
for x in range(tinggi):
    image_3[x, int((lebar) / 2)] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

Garis Menyilang Diagonal :

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

for x in range(lebar):
    # Menghitung Koordinat Y Untuk Diagonal Pertama (Dari Kiri Atas Ke Kanan Bawah)
    y1 = int((tinggi / lebar) * x)
    # Menghitung Koordinat Y Untuk Diagonal Kedua (Dari Kanan Atas Ke Kiri Bawah)
    y2 = int((tinggi / lebar) * (lebar - 1 - x))

    # Memastikan Koordinat Y1 Tidak Melebihi Batas Tinggi Gambar
    if y1 < tinggi:
        image_3[y1, x] = [255, 255, 255]
    # Memastikan Koordinat Y2 Tidak Melebihi Batas Tinggi Gambar
    if y2 < tinggi:
        image_3[y2, x] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

3. Membuat garis horizontal berwarna putih dibagian tengah gambar dengan
panjang tertentu.

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# Menentukan Titik Awal Dan Titik Akhir Koordinat Horizontal (Panjang Tertentu)
awal_x = int(lebar * 0.15)
akhir_x = int(lebar * 0.4)

# Membuat Garis Horizontal Berwarna Putih Di Tengah Gambar Sepanjang Rentang Tertentu
for y in range(awal_x, akhir_x):
    image_3[int((tinggi) / 2), y] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

4. Membuat kotak menggunakan kumpulan pixel warna putih di sembarang tempat
dalam gambar.

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# Menentukan Koordinat Awal Dan Akhir Untuk Sumbu Y Baris
awal_y = int(tinggi * 0.25)
akhir_y = int(tinggi * 0.65)

# Menentukan Koordinat Awal Dan Akhir Untuk Sumbu X Kolom
awal_x = int(lebar * 0.25)
akhir_x = int(lebar * 0.55)

# Membuat Kotak Putih Menggunakan Metode Slicing Array NumPy
image_3[awal_y:akhir_y, awal_x:akhir_x] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```
